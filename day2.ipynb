```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Hours_Studied": [2, 5, 3, 7, 1, 6, 4, 8, 5, 3],
    "Attendance": [60, 90, 75, 95,50,88,80,98,85,70],
    "Previous_Score": [45,70,55,82,40,78,65,90,72,58],
    "Final_Score": [42,76,58,88,35,82,68,94,79,60]
}
df = pd.DataFrame(data)
print(df.head())
print("Shape:",df.shape)
print(df.isnull().sum())

sns.scatterplot(data=df, x="Hours_Studied", y="Final_Score")
plt.title("Study Hours vs Final Score")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data = {
    "Song": ["Shape of You", "Perfect", "Believer",
             "Blinding Lights", "Levitating", "Stay",
             "Peaches", "Memories", "Senorita", "Closer"],
    "Popularity": [95, 90, 85, 98, 92, 88, 87, 80, 89, 93]
}

df = pd.DataFrame(data)

print(df)

sns.barplot(data=df, x="Song", y="Popularity")
plt.title("Spotify Song Popularity Analysis")
plt.xlabel("Songs")
plt.ylabel("Popularity")
plt.xticks(rotation=45)
plt.show()
```

```python
import pandas as pd
from google.colab import files

# Save the DataFrame to an Excel file
excel_filename = 'spotify_popularity.xlsx'
df.to_excel(excel_filename, index=False)
print(f"DataFrame successfully saved to {excel_filename}")

# Trigger browser download of the Excel file
files.download(excel_filename)
```

```python
# Sort the DataFrame by Popularity in descending order
df_sorted = df.sort_values(by='Popularity', ascending=False)
display(df_sorted)
```
